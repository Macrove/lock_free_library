```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import json
with open("bm_stats/l_bm_nopin.json", 'r') as f:
    data = json.load(f)
    l_nopin_df = pd.json_normalize(data["benchmarks"])
with open("bm_stats/lf_bm_nopin.json", 'r') as f:
    data = json.load(f)
    lf_nopin_df = pd.json_normalize(data["benchmarks"])
with open("bm_stats/l_bm_pin.json", 'r') as f:
    data = json.load(f)
    l_pin_df = pd.json_normalize(data["benchmarks"])
with open("bm_stats/lf_bm_pin.json", 'r') as f:
    data = json.load(f)
    lf_pin_df = pd.json_normalize(data["benchmarks"])
```

```python
iterations = l_nopin_df["name"].apply(lambda x: x.split("/")[-1])
plt.plot(iterations, lf_nopin_df['real_time']/iterations.astype(int), label="lock free real time")
plt.plot(iterations, l_nopin_df['real_time']/iterations.astype(int), label="locked real time")
#plt.plot(iterations, lf_nopin_df['cpu_time']/iterations.astype(int), label="lock free cpu time")
#plt.plot(iterations, l_nopin_df['cpu_time']/iterations.astype(int), label="locked cpu time")
plt.xlabel("Num elements pushed by producer. Q cap=1e4")
plt.ylabel("Time taken per element (ns)")
plt.title("SPSC Queue benchmark results. (W/o core pinning)")
plt.yscale("log")
plt.xscale("log")
plt.legend()
plt.show()
```

```python
iterations = l_pin_df["name"].apply(lambda x: x.split("/")[-1])
plt.plot(iterations, lf_pin_df['real_time']/iterations.astype(int), label="lock free real time")
plt.plot(iterations, l_pin_df['real_time']/iterations.astype(int), label="locked real time")
#plt.plot(iterations, lf_pin_df['cpu_time']/iterations.astype(int), label="lock free cpu time")
#plt.plot(iterations, l_pin_df['cpu_time']/iterations.astype(int), label="locked cpu time")
plt.xlabel("Num elements pushed by producer. Q cap=1e4")
plt.ylabel("Time taken per element (ns)")
plt.title("SPSC Queue benchmark results. (With core pinning)")
plt.yscale("log")
plt.xscale("log")
plt.legend()
plt.show()
```
